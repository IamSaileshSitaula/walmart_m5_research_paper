```python
df = pd.read_csv(r"C:\Users\saile\OneDrive\Desktop\research\data\sales_train_validation.csv")
df.head()
```

### 1. state_id = ca 2.store_id = ca_1 3. d_1-d_1913 check for sales =!0

```python
cleaned = df[df['store_id'].isin(['CA_1', 'CA_2', 'CA_3'])]

```

```python
d_cols = [c for c in cleaned.columns if c.startswith('d_')]
```

```python
cleaned['zero_pct'] = (cleaned[d_cols] == 0).mean(axis=1) * 100
top_10_rows = cleaned.nsmallest(12, 'zero_pct')
```

```python
top_10_rows
```

```python
top_10_rows.reset_index(drop=True, inplace = True)
top_10_rows
```

```python
top_10_rows = top_10_rows.drop(columns=['id', 'dept_id', 'cat_id'])
```

```python
top_10_rows
```

```python
# All day columns
d_cols = [c for c in top_10_rows.columns if c.startswith("d_")]
```

```python
long_df = top_10_rows.melt(
    id_vars=["item_id", "store_id"],
    value_vars=d_cols,
    var_name="day",
    value_name="sales"
)
```

```python
calendar_df = pd.read_csv(r"C:\Users\saile\OneDrive\Desktop\research\data\calendar.csv")
```

```python
sellprice_df = pd.read_csv(r"C:\Users\saile\OneDrive\Desktop\research\data\sell_prices.csv")
sellprice_df
```

```python
calendar_df = pd.read_csv(r"C:\Users\saile\OneDrive\Desktop\research\data\calendar.csv")
final_df = long_df.merge(calendar_df, left_on="day", right_on="d", how="left").drop(columns=["d"])

final_df
```

```python
final = final_df.merge(
    sellprice_df,
    on=["item_id", "store_id", "wm_yr_wk"],
    how="left"
)
```

```python
final
```

```python
final = final.drop(columns=['day', 'wm_yr_wk', 'weekday', 'wday', 'month', 'year', 'snap_TX', 'snap_WI'], errors='ignore')
```

```python
final.to_csv(r"C:\Users\saile\OneDrive\Desktop\research\data\final_cleaned.csv", index=False)
print("Saved to data/final_cleaned.csv")
```
